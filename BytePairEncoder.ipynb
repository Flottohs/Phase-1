```python
text_input = "lorem ipsum dolor sit amet, consectetur adipiscing elit, sed do eiusmod tempor incididunt ut labore et dolore magna aliqua."#copied from online

def BytePairEncoder(text_str, vocab_size):
    newchar = 256
    tokens = [int(b) for b in text_str.encode("utf-8")]
    current_vocab_count = len(set(tokens))

    while current_vocab_count < vocab_size:
        pair_counts = {}
        for i in range(len(tokens) - 1):
            pair = (tokens[i], tokens[i+1])
            pair_counts[pair] = pair_counts.get(pair, 0) + 1

        if not pair_counts:
            break

        best_pair = max(pair_counts, key=pair_counts.get)

        new_tokens = []
        i = 0
        while i < len(tokens):
            if i < len(tokens) - 1 and (tokens[i], tokens[i+1]) == best_pair:
                new_tokens.append(newchar)

                i += 2
            else:
                new_tokens.append(tokens[i])
                i += 1

        tokens = new_tokens
        current_vocab_count += 1
        newchar +=1

    return tokens

result = BytePairEncoder(text_input, 40)
print(result)
```
